```python
with open('замок.test', 'r', encoding='utf-8-sig') as file:
    sentences = file.read().splitlines()

print("Количество предложений:", len(sentences))

for sentence in sentences[:10]:
    print(sentence)
```

```python
count_zamok = sum('за`мок' in s.lower() for s in sentences)
count_zamok_lock = sum('замо`к' in s.lower() for s in sentences)

print("Использование зАмок:", count_zamok)
print("Использование замОк:", count_zamok_lock)
```

```python
import re

data = []

for sentence in sentences:
    sentence_lower = sentence.lower()

    if 'за`мок' in sentence_lower:
        label = 0
    elif 'замо`к' in sentence_lower:
        label = 1
    else:
        continue

    clean_sentence = sentence_lower.replace('за`мок', 'замок')
    clean_sentence = clean_sentence.replace('замо`к', 'замок')

    tokens = re.findall(r'[а-яё]+', clean_sentence)

    data.append({
        'tokens': tokens,
        'label': label
    })

print("Подготовлено примеров:", len(data))
print(data[:3])
```

```python
from sklearn.model_selection import train_test_split

train_data, test_data = train_test_split(
    data,
    test_size=0.2,
    random_state=42,
    stratify=[item['label'] for item in data]
)

print("Обучающая выборка:", len(train_data))
print("Тестовая выборка:", len(test_data))
```

```python
from gensim.models import Word2Vec

sentences_train = [item['tokens'] for item in train_data]

w2v_model = Word2Vec(
    sentences=sentences_train,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    sg=1,
    epochs=100,
    seed=42
)

print("Размер словаря:", len(w2v_model.wv))
```

```python
def get_context_vector(tokens, model, window=5):
    target_index = tokens.index('замок')

    left_words = tokens[max(0, target_index-window):target_index]
    right_words = tokens[target_index+1:target_index+1+window]

    def average_vector(words):
        vectors = [
            model.wv[word]
            for word in words
            if word in model.wv
        ]

        if len(vectors) == 0:
            return np.zeros(model.vector_size)

        return np.mean(vectors, axis=0)

    left_vector = average_vector(left_words)
    right_vector = average_vector(right_words)
    all_vector = average_vector(left_words + right_words)

    return np.concatenate([
        left_vector,
        right_vector,
        all_vector
    ])
```

```python
example = train_data[0]

vector = get_context_vector(
    example['tokens'],
    w2v_model
)

print("Размер вектора:", vector.shape)
```

```python
X_train = np.array([
    get_context_vector(item['tokens'], w2v_model)
    for item in train_data
])

y_train = np.array([
    item['label']
    for item in train_data
])

X_test = np.array([
    get_context_vector(item['tokens'], w2v_model)
    for item in test_data
])

y_test = np.array([
    item['label']
    for item in test_data
])

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

classifier = LogisticRegression(
    max_iter=1000,
    class_weight='balanced',
    random_state=42
)

classifier.fit(X_train, y_train)

print("Обучение завершено!")
```

```python
from sklearn.metrics import accuracy_score, classification_report

y_pred = classifier.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Точность модели:", accuracy)

print(classification_report(
    y_test,
    y_pred,
    target_names=['зАмок', 'замОк']
))
```

```python
def predict_zamok(sentence):
    clean_sentence = sentence.lower()

    clean_sentence = clean_sentence.replace('за`мок', 'замок')
    clean_sentence = clean_sentence.replace('замо`к', 'замок')

    tokens = re.findall(r'[а-яё]+', clean_sentence)

    if 'замок' not in tokens:
        return "Слово «замок» не найдено"

    vector = get_context_vector(tokens, w2v_model).reshape(1, -1)

    prediction = classifier.predict(vector)[0]

    if prediction == 0:
        return "зАмок"

    return "замОк"
```

```python
print(predict_zamok("Туристы посетили старый замок"))
print(predict_zamok("Я потерял ключ от замка"))
print(predict_zamok("На двери сломался замок"))
```

```python
import joblib

model_data = {
    'word2vec': w2v_model,
    'classifier': classifier,
    'labels': {
        0: 'зАмок',
        1: 'замОк'
    },
    'window': 5
}

joblib.dump(model_data, 'zamok_model.pkl')

print("Модель сохранена!")
```
